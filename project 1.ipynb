# 2. Student Performance Analyzer 

Create a Student Performance Analyzer that stores academic records and derives 
meaningful insights. 
The system should allow adding students with roll numbers and names, entering marks for 
multiple subjects, and storing them in structured formats. It should calculate individual 
averages, subject-wise toppers, pass/fail status based on defined rules, and overall ranking 
without using built-in sorting functions. 
Marks must be validated, grading rules should differ across subjects, and summary reports 
must be generated using loops and conditional logic. The program must be fully 
menu-driven and modular.

```python
students = []  
pass_rules = {
    "Math": 40,
    "English": 35,
    "Science": 30
}
def add_student():
    roll = input("Enter roll number: ")
    # Check duplicate roll number
    for s in students:
        if s["roll"] == roll:
            print("Roll number already exists!")
            return
    name = input("Enter student name: ")
    students.append({"roll": roll, "name": name, "marks": {}})
    print(f"Student {name} added successfully.")

def enter_marks():
    roll = input("Enter roll number: ")
    student = None
    for s in students:
        if s["roll"] == roll:
            student = s
            break
    if not student:
        print("Student not found!")
        return

    for subject in pass_rules.keys():
        while True:
            try:
                mark = int(input(f"Enter marks for {subject} (0-100): "))
                if 0 <= mark <= 100:
                    student["marks"][subject] = mark
                    break
                else:
                    print("Invalid! Marks must be between 0 and 100.")
            except ValueError:
                print("Invalid input! Enter a number.")

def calculate_average(student):
    marks = student["marks"].values()
    if not marks:
        return 0
    return sum(marks) / len(marks)

def pass_fail(student):
    for subject, pass_mark in pass_rules.items():
        if subject not in student["marks"] or student["marks"][subject] < pass_mark:
            return "Fail"
    return "Pass"

def subject_toppers():
    for subject in pass_rules.keys():
        topper = None
        top_score = -1
        for s in students:
            if subject in s["marks"]:
                if s["marks"][subject] > top_score:
                    top_score = s["marks"][subject]
                    topper = s["name"]
        if topper:
            print(f"Topper in {subject}: {topper} ({top_score})")
        else:
            print(f"No marks entered for {subject} yet.")
def overall_ranking():
    ranking = []
    for s in students:
        avg = calculate_average(s)
        ranking.append((s["roll"], s["name"], avg))
    n = len(ranking)
    for i in range(n):
        for j in range(0, n-i-1):
            if ranking[j][2] < ranking[j+1][2]:
                ranking[j], ranking[j+1] = ranking[j+1], ranking[j]

    print("\n--- Overall Ranking ---")
    for idx, (roll, name, avg) in enumerate(ranking, start=1):
        print(f"{idx}. {name} (Roll {roll}) - Avg: {avg:.2f}")

def summary_report():
    print("\n--- Summary Report ---")
    for s in students:
        avg = calculate_average(s)
        status = pass_fail(s)
        print(f"Roll: {s['roll']}, Name: {s['name']}, Avg: {avg:.2f}, Status: {status}")
def menu():
    while True:
        print("\n--- Student Performance Analyzer ---")
        print("1. Add Student")
        print("2. Enter Marks")
        print("3. Show Summary Report")
        print("4. Show Subject Toppers")
        print("5. Show Overall Ranking")
        print("6. Exit")
        choice = input("Enter choice: ")

        if choice == "1":
            add_student()
        elif choice == "2":
            enter_marks()
        elif choice == "3":
            summary_report()
        elif choice == "4":
            subject_toppers()
        elif choice == "5":
            overall_ranking()
        elif choice == "6":
            print("Exiting program...")
            break
        else:
            print("Invalid choice! Try again.")

if __name__ == "__main__":
    menu()
```

#### 9. Text-Based Decision Game

Create a text-based decision-making game. The game should present players with scenarios 
where choices affect the storyline and score. Multiple story paths and endings must
exist. The system should track decisions, compute scores, and allow replayability.
String processing and conditional branching are core requirements

```python
def play_game():
    score = 0
    decisions = []

    print("\n=== THE LOST EXPEDITION ===")
    print("You are an explorer searching for a legendary artifact.\n")

    # SCENE 1
    choice1 = input(
        "You arrive at a fork in the jungle.\n"
        "Do you go LEFT toward the river or RIGHT into the cave? (left/right): "
    ).lower()

    if choice1 == "left":
        score += 10
        decisions.append("river")
        print("\nYou follow the river and find fresh water. (+10 points)")
    elif choice1 == "right":
        score += 5
        decisions.append("cave")
        print("\nYou enter the cave carefully. It's dark but safe. (+5 points)")
    else:
        print("\nInvalid choice. You hesitate and lose time.")
        score -= 5

    # SCENE 2
    choice2 = input(
        "\nYou encounter a wounded traveler.\n"
        "Do you HELP them or IGNORE them? (help/ignore): "
    ).lower()

    if choice2 == "help":
        score += 15
        decisions.append("helped traveler")
        print("\nThe traveler thanks you and gives you a map. (+15 points)")
    elif choice2 == "ignore":
        score -= 10
        decisions.append("ignored traveler")
        print("\nYou move on, but guilt slows you down. (-10 points)")
    else:
        print("\nYou do nothing. Time passes.")
        score -= 5

    # SCENE 3
    choice3 = input(
        "\nYou reach the artifact chamber.\n"
        "Do you TAKE the artifact or LEAVE it? (take/leave): "
    ).lower()

    if choice3 == "take":
        if "helped traveler" in decisions:
            score += 30
            print("\nThe map helps you escape safely! (+30 points)")
            ending = "SUCCESS ENDING"
        else:
            score -= 20
            print("\nA trap is triggered! You barely escape. (-20 points)")
            ending = "DANGEROUS ENDING"
    elif choice3 == "leave":
        score += 10
        print("\nYou leave peacefully, wiser than before. (+10 points)")
        ending = "PEACEFUL ENDING"
    else:
        print("\nYou hesitate too long. The chamber collapses.")
        score -= 15
        ending = "FAILURE ENDING"

    # FINAL RESULTS
    print("\n=== GAME OVER ===")
    print(f"Ending: {ending}")
    print(f"Final Score: {score}")

def main():
    while True:
        play_game()
        replay = input("\nWould you like to play again? (yes/no): ").lower()
        if replay != "yes":
            print("\nThanks for playing!")
            break

main()
```
