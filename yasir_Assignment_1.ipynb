```python
# Q1
item1_price = int(input("Enter Item1 price: "))
item1_quantity = int(input("Enter Item1 quantity: "))
item2_price = int(input("Enter Item2 price: "))
item2_quantity = int(input("Enter Item2 quantity: "))
t= (item1_price * item1_quantity) + (item2_price * item2_quantity)
print("Total Bill:",t)
```

```python
# Q2
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print("Temperature in Fahrenheit:", fahrenheit)
```

```python
# Q3
basic_salary = float(input("Enter basic salary: "))
hra = 0.20 * basic_salary
da = 0.10 * basic_salary
net_salary = basic_salary + hra + da
print("Net Salary:", net_salary)
```

```python
# Q4
P = float(input("Principal: "))
R = float(input("Rate of Interest: "))
T = float(input("Time (in years): "))
SI = (P * R * T) / 100
print("Simple Interest =", SI)
```

```python
# Q5
total_minutes = int(input("Enter total minutes: "))
hours = total_minutes // 60 
minutes = total_minutes % 60 
print(f"{hours} hours and {minutes} minutes")
```

```python
# Q6
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))
bmi = weight / (height * height)
print("BMI = {:.2f}".format(bmi))
```

```python
# Q7
original_price = float(input("Enter the original price: "))
discount = original_price * 0.10
discounted_price = original_price - discount
print("Discounted Price:", discounted_price)
```

```python
# Q8
import math
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print("Area of the triangle: {:.2f}".format(area))
```

```python
# Q9
length = float(input("Enter wall length (in meters): "))
height = float(input("Enter wall height (in meters): "))
window_length = 1.5
window_height = 1.2
window_area = window_length * window_height
num_windows = 2
rate_per_sq_m = 25  
wall_area = length * height
paint_area = wall_area - (num_windows * window_area)
total_cost = paint_area * rate_per_sq_m
print("Total area to be painted: {:.1f} sq.m".format(paint_area))
print("Total painting cost: ₹{:.1f}".format(total_cost))
```

```python
# Q10
units = float(input("Enter number of units consumed: "))
rate_per_unit = 5.2   
fixed_charge = 100   
bill_amount = (units * rate_per_unit) + fixed_charge
print("Electricity Bill: ₹{:.1f}".format(bill_amount))
```
